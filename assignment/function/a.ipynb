```python
#fixed length argument

def add(a, b):
    return a+b
```

```python
add(90, 90)
```

```python
#variable length argument
def add(*numbers):
    result = 0 #identity operation, read more on abstract orgibral
    for number in numbers:
        result = result + number
    return result
```

```python
add(23, 67)
```

```python
#factorial recursion

def factorial(n:int)-> int:
    if n ==0: return 1

    return n * factorial(n - 1)
```

```python
factorial(3)
```
